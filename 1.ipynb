# Задача 1: Комплексные числа



## Математическая база

Комплексное число $z$ можно представить в трёх основных формах:

| Форма | Запись | Описание |
| :--- | :--- | :--- |
| **Алгебраическая** | $z = a + bi$ | $a, b \in \mathbb{R}$, $i^2 = -1$ |
| **Тригонометрическая** | $z = r(\cos\varphi + i\sin\varphi)$ | $r$ — модуль, $\varphi$ — аргумент |
| **Показательная** | $z = r e^{i\varphi}$ | Следует из формулы Эйлера $e^{i\varphi} = \cos\varphi + i\sin\varphi$ |

### Ключевые формулы

1. **Модуль комплексного числа**:
   $$r = |z| = \sqrt{a^2 + b^2}$$

2. **Аргумент комплексного числа** (главное значение $\varphi \in (-\pi, \pi]$):
   $$\varphi = \arg(z) = \operatorname{arctg}(\frac{b}{a})$$
   где $\cos\varphi = \frac{a}{r}$, $\sin\varphi = \frac{b}{r}$.

3. **Умножение и деление** (в показательной форме):
   $$z_1 \cdot z_2 = r_1 r_2 \, e^{i(\varphi_1 + \varphi_2)}$$
   $$\frac{z_1}{z_2} = \frac{r_1}{r_2} \, e^{i(\varphi_1 - \varphi_2)}$$

4. **Формула Муавра** (возведение в степень $n \in \mathbb{N}$):
   $$z^n = \left[ r(\cos\varphi + i\sin\varphi) \right]^n = r^n (\cos(n\varphi) + i\sin(n\varphi)) = r^n e^{in\varphi}$$


```python
import sympy as sp

def show_forms(z, name="z"):
    z = sp.simplify(z)
    a = sp.re(z)
    b = sp.im(z)
    r = sp.sqrt(a**2 + b**2)
    phi = sp.arg(z)          # аргумент в (-π, π]

    print(f"  {name} = {z}")
    print(f"  Алгебраическая:     z = {a} + ({b})i")
    print(f"  Модуль r = {r}  ≈  {float(r):.6f}")
    print(f"  Аргумент φ = {phi}  ≈  {float(phi):.6f} рад")
    print(f"  Тригонометрическая: z = {r}(cos({phi}) + i·sin({phi}))")
    print(f"  Показательная:      z = {r}·e^(i·{phi})")


i = sp.I
sqrt3 = sp.sqrt(3)

za = (5 + i) * (7 - 6*i) / (3 + i)
show_forms(za, "a)")

zb = (1 + i)**5 / (1 - i)**3
show_forms(zb, "b)")

zc = ((-1 + i*sqrt3)**15) / (1 - i)**20 + \
     ((-1 - i*sqrt3)**15) / (1 + i)**20
show_forms(zc, "c)")
```
