# Задача 6: Корни многочлена 11-й степени



## Математическая база

### Основная теорема алгебры

Всякий многочлен степени $n \geq 1$ с комплексными коэффициентами имеет ровно $n$ корней в $\mathbb{C}$ (с учётом кратности).

### Свойства корней многочленов с вещественными коэффициентами

Если все коэффициенты вещественны, то комплексные корни приходят **сопряжёнными парами**: если $z$ — корень, то $\bar{z}$ тоже корень.

```python
import numpy as np


coeffs = [1, 1, 0, 1, -2, 0, -1, 1, 0, 2, 2, 2]

# Находим все 11 корней
roots = np.roots(coeffs)

print("  Все 11 корней многочлена p(x)")
for k, z in enumerate(roots, 1):
    print(f"  z_{k:2d} = {z.real:10.6f} + {z.imag:10.6f}i   |z| = {abs(z):.6f}")


sum_moduli = np.sum(np.abs(roots))
print(f"  1) Сумма модулей: |z₁| + ... + |z₁₁| = {sum_moduli:.6f}")


max_re_idx = np.argmax(roots.real)
max_re = roots[max_re_idx].real
max_re_root = roots[max_re_idx]
print(f"  2) Максимальная действительная часть:")
print(f"     max Re(z_k) = {max_re:.6f}")
print(f"     Соответствующий корень: z = {max_re_root.real:.6f} + {max_re_root.imag:.6f}i")
```
